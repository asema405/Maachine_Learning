```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

print("Китепканалар ийгиликтүү жүктөлдү!")
```

```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"

df = pd.read_csv(url)

print("Датасет ийгиликтүү жүктөлдү!")
print("Алгачкы 5 сап:")
display(df.head())
```

```python
print("Таблицанын өлчөмү:", df.shape)
print("Саптардын саны:", df.shape[0])
print("Мамычалардын саны:", df.shape[1])
```

```python
df.info()
```

```python
display(df.describe())
```

```python
print("Мамычалардын маалымат түрлөрү:")
display(df.dtypes)
```

```python
print("Ар бир мамычадагы бош маанилердин саны:")
display(df.isnull().sum())

print("Жалпы бош маанилердин саны:", df.isnull().sum().sum())
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    df["total_bill"],
    bins=20,
    edgecolor="black"
)

plt.title("Ресторандагы эсептердин бөлүштүрүлүшү")
plt.xlabel("Эсептин суммасы")
plt.ylabel("Эсептердин саны")

plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
day_counts = df["day"].value_counts().reindex(
    ["Thur", "Fri", "Sat", "Sun"]
)

plt.figure(figsize=(8, 5))

plt.bar(
    day_counts.index,
    day_counts.values,
    edgecolor="black"
)

plt.title("Аптанын күндөрү боюнча эсептердин саны")
plt.xlabel("Күн")
plt.ylabel("Эсептердин саны")

plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

## Данные и наблюдения

2. В датасете представлены как небольшие, так и крупные ресторанные счета. Гистограмма позволяет увидеть, в каких диапазонах суммы встречаются чаще.

3. В таблице есть числовые и категориальные признаки. Это позволяет анализировать не только суммы счетов и чаевых, но и различия между днями недели, временем посещения и другими категориями.
